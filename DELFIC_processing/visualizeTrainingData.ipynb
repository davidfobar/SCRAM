```python
import numpy as np
import matplotlib.pyplot as plt

#open a sample file
x = np.load('../data/Y100H0SS4DS50SA15BA5441_X.npy', allow_pickle=True)

print(x.shape)
```

```python
# X is of form [nSim, timestep, nDet, features]
#the 6 features are: x, y, dPos, dose, dDose, detNum

#plot the final x and y positions of the detectors
sim = 0
plt.scatter(x[sim, -1, :, 0], x[sim, -1, :, 1])
plt.show()

# plot the dose of the first 10 detectors across time
for i in range(10):
    plt.plot(x[sim, :, i, 3])
plt.show()
```
